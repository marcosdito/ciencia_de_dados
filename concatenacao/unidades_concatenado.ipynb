```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_unidades_curso = pd.read_csv('../unidades_academicas/unidades_e_cursos.csv', sep=';')
df_tccs = pd.read_csv('../TCC/tccs_tratado3.0.csv', sep=';')
df_pesquisa = pd.read_csv('../projetos-de-pesquisa-limpo2.0.csv')
```

```python
df_unidades_curso.info()
```

```python
df_tccs.info()
```

```python
df_pesquisa.info()
```

```python
for df in [df_unidades_curso, df_tccs]:
    df["curso"] = (df["curso"] #padronizar
                   .astype(str)
                   .str.strip()
                   .str.lower()
                   .str.normalize("NFKD")
                   .str.encode("ascii", errors="ignore")
                   .str.decode("utf-8"))

df_temp = pd.merge(df_tccs, df_unidades_curso, on="curso", how="outer")

df_temp.head()
```

```python
df_temp.info()
```

```python
df_temp2 = pd.merge(df_pesquisa, df_unidades_curso, on="unidade", how="outer")

df_temp2.head(50)
```

```python
df_temp2.info()
```

```python
df_final = pd.concat([df_temp, df_temp2], axis=0, join="outer", ignore_index=True)
```
