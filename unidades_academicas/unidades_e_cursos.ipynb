```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_cursos = pd.read_csv('cursos-ufrn.csv', sep=';')
```

```python
df_cursos.info()
```

```python
df_cursos = df_cursos.drop(columns=['id_coordenador', 'coordenador', 'situacao_curso', 'nivel_ensino', 'grau_academico', 'modalidade_educacao', 'area_conhecimento', 'tipo_oferta', 'turno', 'tipo_ciclo_formacao', 'municipio', 'campus', 'website', 'data_funcionamento', 'codigo_inep', 'dou', 'portaria_reconhecimento', 'convenio_academico'])
```

```python
#Mudando o nome pra padronizar com os outros tccs_tratado2.0.csv e projetos-de-pesquisa-limpo.csv
df_cursos = df_cursos.rename(columns={'nome': 'curso'})
df_cursos = df_cursos.rename(columns={'id_unidade_responsavel': 'id_unidade'})
df_cursos = df_cursos.rename(columns={'unidade_responsavel': 'unidade'})
```

```python
df_cursos.info()
```

```python
df_cursos = df_cursos.drop('id_curso', axis=1)
```

```python
df_cursos = df_cursos.drop('id_unidade', axis=1)
```

```python
df_cursos = df_cursos[df_cursos['unidade'] != 'CENTRO DE  ENSINO SUPERIOR DO SERIDÓ']
```

```python
df_cursos = df_cursos.drop_duplicates()
```

```python
 df_cursos.to_csv(
     'unidades_e_cursos.csv',
     sep=';',
     encoding='utf-8',
     index=False
 )
```
